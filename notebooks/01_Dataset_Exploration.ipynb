### Data Cleaning

```python
data_clean= data.copy()
```

```python
data_clean['TotalCharges']=data_clean['TotalCharges'].replace(' ',np.nan)
```

```python
data_clean['TotalCharges'] = pd.to_numeric(data_clean['TotalCharges'], errors='coerce')
```

```python
data_clean["TotalCharges"].dtype
```

```python
data_clean[data_clean['TotalCharges'].isnull()][['tenure','MonthlyCharges', 'Contract', 'Churn']]
```

Observation      
11 of the customers that have total amount null is because they are new.




```python
## Fill missing values in TotalCharges with 0 as these customers haven't been charged yet. 
# This is a business decision and can be changed based on the business requirements.
data_clean['TotalCharges']= data_clean['TotalCharges'].fillna(0)
```
